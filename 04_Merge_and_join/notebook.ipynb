```python
import pandas as pd
```

```python
customers = pd.read_csv("data/customer.csv")

products = pd.read_csv("data/product.csv")

orders = pd.read_csv("data/order.csv")
```

```python
orders_customers = pd.merge(
    orders,
    customers,
    on="CustomerID"
)
```

```python
sales = pd.merge(
    orders_customers,
    products,
    on="ProductID"
)
```

```python
sales.head()
```

```python
sales["Revenue"] = (
    sales["Quantity"] *
    sales["Price"]
)
```

```python
sales.to_csv(
    "output/sales_report.csv",
    index=False
)
```

```python
inner_join = pd.merge(
    orders,
    customers,
    on="CustomerID",
    how="inner"
)
```

```python
left_join = pd.merge(
    orders,
    customers,
    on="CustomerID",
    how="left"
)
```

```python
right_join = pd.merge(
    orders,
    customers,
    on="CustomerID",
    how="right"
)
```

```python
outer_join = pd.merge(
    orders,
    customers,
    on="CustomerID",
    how="outer"
)
```

```python
print(inner_join.shape)

print(left_join.shape)

print(right_join.shape)

print(outer_join.shape)
```

```python
inner_join.to_csv("output/inner_join.csv", index=False)

left_join.to_csv("output/left_join.csv", index=False)

right_join.to_csv("output/right_join.csv", index=False)

outer_join.to_csv("output/outer_join.csv", index=False)
```
