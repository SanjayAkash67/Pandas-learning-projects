```python
# ============================================================
# PHASE 5 → FINAL — OLIST COMPLETE PANDAS ANALYSIS
# ============================================================

print("=" * 70)
print("PHASE 5 — BUSINESS ANALYSIS")
print("=" * 70)

# ------------------------------------------------------------
# 1. ORDER STATUS ANALYSIS
# ------------------------------------------------------------

print("\n--- Order Status Count ---")
print(orders["order_status"].value_counts())

print("\n--- Order Status Percentage ---")
print(
    (orders["order_status"].value_counts(normalize=True) * 100)
    .round(2)
)


# ------------------------------------------------------------
# 2. PAYMENT ANALYSIS
# ------------------------------------------------------------

print("\n--- Payment Types ---")
print(order_payments["payment_type"].value_counts())

print("\n--- Payment Statistics ---")
print(
    order_payments[
        ["payment_value", "payment_installments"]
    ].describe()
)

print("\nTotal Payment Value:",
      round(order_payments["payment_value"].sum(), 2))

print("Average Payment Value:",
      round(order_payments["payment_value"].mean(), 2))


# ------------------------------------------------------------
# 3. PRODUCT ANALYSIS
# ------------------------------------------------------------

print("\n--- Total Unique Products ---")
print(products["product_id"].nunique())

print("\n--- Top 10 Product Categories ---")
print(
    products["product_category_name"]
    .value_counts()
    .head(10)
)


# ------------------------------------------------------------
# 4. SALES / REVENUE CALCULATION
# ------------------------------------------------------------

order_items["revenue"] = (
    order_items["price"] +
    order_items["freight_value"]
)

print("\n--- Revenue Statistics ---")
print(
    order_items["revenue"].describe()
)

print("\nTotal Revenue:",
      round(order_items["revenue"].sum(), 2))

print("\n--- Top 10 Highest Value Items ---")
print(
    order_items.nlargest(
        10,
        "revenue"
    )
)


# ============================================================
# PHASE 6 — MERGE / CREATE MASTER DATASET
# ============================================================

print("\n" + "=" * 70)
print("PHASE 6 — MERGING DATASETS")
print("=" * 70)

# Orders + Customers
sales = orders.merge(
    customers,
    on="customer_id",
    how="left"
)

# + Order Items
sales = sales.merge(
    order_items,
    on="order_id",
    how="left"
)

# + Products
sales = sales.merge(
    products,
    on="product_id",
    how="left"
)

# + Sellers
sales = sales.merge(
    sellers,
    on="seller_id",
    how="left"
)

# + English category names
sales = sales.merge(
    category_translation,
    on="product_category_name",
    how="left"
)

print("\nMaster Dataset Shape:")
print(sales.shape)

print("\nMaster Dataset Info:")
sales.info()

print("\nMaster Dataset Preview:")
display(sales.head())


# ============================================================
# PHASE 7 — CUSTOMER ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("PHASE 7 — CUSTOMER ANALYSIS")
print("=" * 70)

print("\n--- Customers by State ---")
print(
    sales["customer_state"]
    .value_counts()
    .head(10)
)

print("\n--- Customers by City ---")
print(
    sales["customer_city"]
    .value_counts()
    .head(10)
)


# ============================================================
# PHASE 8 — CATEGORY ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("PHASE 8 — CATEGORY ANALYSIS")
print("=" * 70)

category_sales = (
    sales
    .groupby("product_category_name_english")
    .agg(
        Orders=("order_id", "nunique"),
        Items=("order_item_id", "count"),
        Revenue=("revenue", "sum")
    )
    .reset_index()
    .sort_values(
        "Revenue",
        ascending=False
    )
)

print("\n--- Top 10 Categories by Revenue ---")
display(
    category_sales.head(10)
)

print("\n--- Top 10 Categories by Orders ---")
display(
    category_sales
    .sort_values(
        "Orders",
        ascending=False
    )
    .head(10)
)


# ============================================================
# PHASE 9 — SELLER ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("PHASE 9 — SELLER ANALYSIS")
print("=" * 70)

seller_sales = (
    sales
    .groupby(
        ["seller_id", "seller_state"]
    )
    .agg(
        Orders=("order_id", "nunique"),
        Items=("order_item_id", "count"),
        Revenue=("revenue", "sum")
    )
    .reset_index()
    .sort_values(
        "Revenue",
        ascending=False
    )
)

print("\n--- Top 10 Sellers ---")
display(
    seller_sales.head(10)
)


# ============================================================
# PHASE 10 — TIME SERIES ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("PHASE 10 — TIME SERIES ANALYSIS")
print("=" * 70)

sales["order_purchase_timestamp"] = pd.to_datetime(
    sales["order_purchase_timestamp"],
    errors="coerce"
)

sales["year"] = (
    sales["order_purchase_timestamp"].dt.year
)

sales["month"] = (
    sales["order_purchase_timestamp"].dt.month
)

sales["month_name"] = (
    sales["order_purchase_timestamp"]
    .dt.month_name()
)

sales["year_month"] = (
    sales["order_purchase_timestamp"]
    .dt.to_period("M")
)

# Monthly Revenue
monthly_sales = (
    sales
    .set_index("order_purchase_timestamp")
    ["revenue"]
    .resample("ME")
    .sum()
    .reset_index()
)

monthly_sales.columns = [
    "Month",
    "Revenue"
]

print("\n--- Monthly Revenue ---")
display(monthly_sales)

# Monthly Orders
monthly_orders = (
    sales
    .set_index("order_purchase_timestamp")
    ["order_id"]
    .resample("ME")
    .nunique()
    .reset_index()
)

monthly_orders.columns = [
    "Month",
    "Orders"
]

print("\n--- Monthly Orders ---")
display(monthly_orders)

# Highest revenue month
print("\nHighest Revenue Month:")
display(
    monthly_sales.loc[
        monthly_sales["Revenue"].idxmax()
    ]
)


# ============================================================
# PHASE 11 — PIVOT TABLES
# ============================================================

print("\n" + "=" * 70)
print("PHASE 11 — PIVOT TABLES")
print("=" * 70)

state_category = pd.pivot_table(
    sales,
    values="revenue",
    index="customer_state",
    columns="product_category_name_english",
    aggfunc="sum",
    fill_value=0
)

print("\n--- State × Category Revenue Pivot ---")
display(
    state_category.head()
)

# Payment × Order Status
payment_summary = pd.crosstab(
    orders["order_status"],
    order_payments["payment_type"]
)

print("\n--- Payment Type × Order Status ---")
display(payment_summary)


# ============================================================
# PHASE 12 — STRING OPERATIONS
# ============================================================

print("\n" + "=" * 70)
print("PHASE 12 — STRING OPERATIONS")
print("=" * 70)

# Clean customer city
sales["customer_city"] = (
    sales["customer_city"]
    .astype("string")
    .str.strip()
    .str.lower()
)

# Clean category
sales["product_category_name_english"] = (
    sales["product_category_name_english"]
    .astype("string")
    .str.strip()
    .str.lower()
)

print("\n--- Categories containing 'bed' ---")

bed_categories = sales[
    sales["product_category_name_english"]
    .str.contains(
        "bed",
        na=False
    )
]

display(
    bed_categories[
        ["product_category_name_english"]
    ]
    .drop_duplicates()
    .head(20)
)


# ============================================================
# PHASE 13 — DELIVERY ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("PHASE 13 — DELIVERY ANALYSIS")
print("=" * 70)

sales["order_delivered_customer_date"] = pd.to_datetime(
    sales["order_delivered_customer_date"],
    errors="coerce"
)

sales["order_estimated_delivery_date"] = pd.to_datetime(
    sales["order_estimated_delivery_date"],
    errors="coerce"
)

sales["delivery_days"] = (
    sales["order_delivered_customer_date"]
    - sales["order_purchase_timestamp"]
).dt.days

sales["delivery_difference"] = (
    sales["order_estimated_delivery_date"]
    - sales["order_delivered_customer_date"]
).dt.days

print("\nAverage Delivery Days:")
print(
    round(
        sales["delivery_days"].mean(),
        2
    )
)

print("\nFastest Delivery:")
print(
    sales["delivery_days"].min()
)

print("\nSlowest Delivery:")
print(
    sales["delivery_days"].max()
)

print("\nDelivery Statistics:")
display(
    sales["delivery_days"].describe()
)


# ============================================================
# PHASE 14 — FINAL BUSINESS QUESTIONS
# ============================================================

print("\n" + "=" * 70)
print("PHASE 14 — FINAL BUSINESS QUESTIONS")
print("=" * 70)


# Q1 — Total Revenue
total_revenue = sales["revenue"].sum()

print("\nQ1. Total Revenue:")
print(round(total_revenue, 2))


# Q2 — Top Category
top_category = (
    category_sales
    .iloc[0]
)

print("\nQ2. Top Category by Revenue:")
print(top_category)


# Q3 — Top Seller
top_seller = seller_sales.iloc[0]

print("\nQ3. Top Seller:")
print(top_seller)


# Q4 — State with highest revenue
state_revenue = (
    sales
    .groupby("customer_state")["revenue"]
    .sum()
    .sort_values(
        ascending=False
    )
)

print("\nQ4. Top State by Revenue:")
print(state_revenue.head(10))


# Q5 — Average Order Value
order_values = (
    sales
    .groupby("order_id")["revenue"]
    .sum()
)

average_order_value = order_values.mean()

print("\nQ5. Average Order Value:")
print(round(average_order_value, 2))


# Q6 — Most common payment type
common_payment = (
    order_payments["payment_type"]
    .value_counts()
    .idxmax()
)

print("\nQ6. Most Common Payment Type:")
print(common_payment)


# Q7 — Highest payment value by payment type
payment_value_summary = (
    order_payments
    .groupby("payment_type")["payment_value"]
    .sum()
    .sort_values(
        ascending=False
    )
)

print("\nQ7. Payment Type by Total Value:")
print(payment_value_summary)


# Q8 — Highest revenue month
highest_revenue_month = (
    monthly_sales
    .loc[
        monthly_sales["Revenue"].idxmax()
    ]
)

print("\nQ8. Highest Revenue Month:")
print(highest_revenue_month)


# Q9 — Highest order month
highest_order_month = (
    monthly_orders
    .loc[
        monthly_orders["Orders"].idxmax()
    ]
)

print("\nQ9. Highest Order Month:")
print(highest_order_month)


# Q10 — Category with highest average item price
category_avg_price = (
    sales
    .groupby(
        "product_category_name_english"
    )["price"]
    .mean()
    .sort_values(
        ascending=False
    )
)

print("\nQ10. Category with Highest Average Item Price:")
print(category_avg_price.head(10))


# Q11 — State with highest average order value
state_order_value = (
    sales
    .groupby(
        ["customer_state", "order_id"]
    )["revenue"]
    .sum()
    .groupby(level=0)
    .mean()
    .sort_values(
        ascending=False
    )
)

print("\nQ11. State with Highest Average Order Value:")
print(state_order_value.head(10))


# Q12 — Delivered order percentage
total_orders = orders["order_id"].nunique()

delivered_orders = orders[
    orders["order_status"] == "delivered"
]["order_id"].nunique()

delivered_percentage = (
    delivered_orders /
    total_orders *
    100
)

print("\nQ12. Delivered Orders %:")
print(round(delivered_percentage, 2))


# Q13 — Cancelled order percentage
cancelled_orders = orders[
    orders["order_status"] == "canceled"
]["order_id"].nunique()

cancelled_percentage = (
    cancelled_orders /
    total_orders *
    100
)

print("\nQ13. Cancelled Orders %:")
print(round(cancelled_percentage, 2))


# Q14 — Average delivery time
print("\nQ14. Average Delivery Time:")
print(
    round(
        sales["delivery_days"].mean(),
        2
    )
)


# Q15 — Category with longest average delivery time
category_delivery = (
    sales
    .groupby(
        "product_category_name_english"
    )["delivery_days"]
    .mean()
    .sort_values(
        ascending=False
    )
)

print("\nQ15. Categories with Longest Average Delivery:")
print(category_delivery.head(10))


# ============================================================
# PHASE 15 — FINAL REPORT EXPORT
# ============================================================

print("\n" + "=" * 70)
print("PHASE 15 — EXPORTING FINAL REPORT")
print("=" * 70)

# CSV reports
category_sales.to_csv(
    "output/category_sales.csv",
    index=False
)

seller_sales.to_csv(
    "output/seller_sales.csv",
    index=False
)

monthly_sales.to_csv(
    "output/monthly_sales.csv",
    index=False
)

monthly_orders.to_csv(
    "output/monthly_orders.csv",
    index=False
)

state_revenue.reset_index(
    name="Revenue"
).to_csv(
    "output/state_revenue.csv",
    index=False
)

# Final Excel workbook
with pd.ExcelWriter(
    "output/olist_analysis_report.xlsx"
) as writer:

    category_sales.to_excel(
        writer,
        sheet_name="Category Sales",
        index=False
    )

    seller_sales.to_excel(
        writer,
        sheet_name="Seller Sales",
        index=False
    )

    monthly_sales.to_excel(
        writer,
        sheet_name="Monthly Revenue",
        index=False
    )

    monthly_orders.to_excel(
        writer,
        sheet_name="Monthly Orders",
        index=False
    )

    state_revenue.reset_index(
        name="Revenue"
    ).to_excel(
        writer,
        sheet_name="State Revenue",
        index=False
    )

    state_category.to_excel(
        writer,
        sheet_name="State Category"
    )

    payment_summary.to_excel(
        writer,
        sheet_name="Payment Summary"
    )

    # Keep Excel file manageable
    sales.head(10000).to_excel(
        writer,
        sheet_name="Sample Master Data",
        index=False
    )

print("\n" + "=" * 70)
print("✅ PANDAS FINAL PROJECT COMPLETED")
print("=" * 70)

print("\nGenerated files:")
print("output/category_sales.csv")
print("output/seller_sales.csv")
print("output/monthly_sales.csv")
print("output/monthly_orders.csv")
print("output/state_revenue.csv")
print("output/olist_analysis_report.xlsx")
```
