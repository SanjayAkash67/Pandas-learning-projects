```python
def categorize_review(review):

    review = review.lower()

    if "excellent" in review:
        return "Excellent"

    elif "good" in review:
        return "Good"

    elif "average" in review:
        return "Average"

    elif "late" in review:
        return "Negative"

    else:
        return "Other"
```

```python
df["ReviewCategory"] = df["Review"].apply(
    categorize_review
)
```

```python
df["WordCount"] = df["Review"].str.split().str.len()
```

```python
df["CharacterCount"] = df["Review"].str.len()
```

```python
negative_keywords = [
    "late",
    "bad",
    "poor",
    "damaged"
]
```

```python
pattern = "|".join(negative_keywords)
```

```python
pattern
```

```python
df["HasNegativeKeyword"] = df["Review"].str.contains(
    pattern,
    case=False,
    na=False
)
```

```python
positive_keywords = [
    "excellent",
    "good",
    "great"
]
```

```python
positive_pattern = "|".join(positive_keywords)
```

```python
df["HasPositiveKeyword"] = df["Review"].str.contains(
    positive_pattern,
    case=False,
    na=False
)
```

```python
df["HasPositiveKeyword"].sum()
```

```python
df["HasNegativeKeyword"].sum()
```

```python
def classify_sentiment(row):

    if row["HasPositiveKeyword"] and not row["HasNegativeKeyword"]:
        return "Positive"

    elif row["HasNegativeKeyword"] and not row["HasPositiveKeyword"]:
        return "Negative"

    elif row["HasPositiveKeyword"] and row["HasNegativeKeyword"]:
        return "Mixed"

    else:
        return "Neutral"
```

```python
df["Sentiment"] = df.apply(
    classify_sentiment,
    axis=1
)
```
