```python
df["Average"] = (
    df["Math"] +
    df["Science"] +
    df["English"]
) / 3
```

```python
df["Total"] = (
    df["Math"] +
    df["Science"] +
    df["English"]
)
```

```python
def grade(avg):
    if avg >= 90:
        return "A"
    elif avg >= 80:
        return "B"
    elif avg >= 70:
        return "C"
    else:
        return "D"

df["Grade"] = df["Average"].apply(grade)
```
