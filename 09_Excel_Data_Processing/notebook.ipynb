```python
sales = pd.read_excel(
    excel_file,
    sheet_name="Sales"
)
```

```python
sales_details = pd.merge(
    sales,
    employees,
    on="EmployeeID",
    how="left"
)
```

```python
sales_details.head()
```

```python
department_sales = (
    sales_details
    .groupby("Department")["Revenue"]
    .sum()
    .reset_index()
)
```

```python
department_sales.columns = [
    "Department",
    "Revenue"
]
```

```python
department_report = pd.merge(
    department_sales,
    targets,
    on="Department",
    how="left"
)
```

```python
department_report["AchievementPercent"] = (
    department_report["Revenue"]
    / department_report["MonthlyTarget"]
    * 100
)
```

```python
department_report["AchievementPercent"] = (
    department_report["AchievementPercent"]
    .round(2)
)
```

```python
def target_status(row):

    if row["Revenue"] >= row["MonthlyTarget"]:
        return "Target Achieved"

    return "Below Target"
```

```python
department_report["Status"] = (
    department_report.apply(
        target_status,
        axis=1
    )
)
```

```python
department_report
```

```python
product_report = (
    sales_details
    .groupby("Product")
    .agg(
        TotalQuantity=("Quantity", "sum"),
        TotalRevenue=("Revenue", "sum"),
        AveragePrice=("Price", "mean")
    )
    .reset_index()
)
```

```python
column_name=("source_column", "function")
```

```python
employee_report = (
    sales_details
    .groupby(
        ["EmployeeID", "Name", "Department"]
    )
    .agg(
        Orders=("OrderID", "count"),
        TotalQuantity=("Quantity", "sum"),
        TotalRevenue=("Revenue", "sum")
    )
    .reset_index()
)
```

```python
employee_report = employee_report.sort_values(
    "TotalRevenue",
    ascending=False
)
```

```python
with pd.ExcelWriter(
    "output/company_analysis_report.xlsx"
) as writer:

    sales_details.to_excel(
        writer,
        sheet_name="Sales Details",
        index=False
    )

    department_report.to_excel(
        writer,
        sheet_name="Department Report",
        index=False
    )

    product_report.to_excel(
        writer,
        sheet_name="Product Report",
        index=False
    )

    employee_report.to_excel(
        writer,
        sheet_name="Employee Report",
        index=False
    )
```

```python
city_report = (
    sales_details
    .groupby("City")
    .agg(
        Orders=("OrderID", "count"),
        TotalQuantity=("Quantity", "sum"),
        TotalRevenue=("Revenue", "sum")
    )
    .reset_index()
)
```

```python
city_report = city_report.sort_values(
    "TotalRevenue",
    ascending=False
)
```

```python
category_report = (
    sales_details
    .groupby("Category")
    .agg(
        Orders=("OrderID", "count"),
        TotalQuantity=("Quantity", "sum"),
        TotalRevenue=("Revenue", "sum")
    )
    .reset_index()
)
```

```python
category_report = category_report.sort_values(
    "TotalRevenue",
    ascending=False
)
```
