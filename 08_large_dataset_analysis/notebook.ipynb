```python
import pandas as pd

FILE_PATH = "data/transactions.csv"
CHUNK_SIZE = 10_000
def analyze_transactions(file_path, chunk_size):

    total_transactions = 0
    total_revenue = 0

    category_results = []
    city_results = []
    customer_results = []

    for chunk in pd.read_csv(
        file_path,
        chunksize=chunk_size
    ):

        total_transactions += len(chunk)

        total_revenue += chunk["Revenue"].sum()

        category_results.append(
            chunk.groupby("Category")["Revenue"].sum()
        )

        city_results.append(
            chunk.groupby("City")["Revenue"].sum()
        )

        customer_results.append(
            chunk.groupby("CustomerID")["Revenue"].sum()
        )

    return (
        total_transactions,
        total_revenue,
        category_results,
        city_results,
        customer_results
    )
```

```python
results = analyze_transactions(
    FILE_PATH,
    CHUNK_SIZE
)
```

```python
(
    total_transactions,
    total_revenue,
    category_results,
    city_results,
    customer_results
) = results
```

```python
print(
    "Total Transactions:",
    total_transactions
)

print(
    "Total Revenue:",
    total_revenue
)
```

```python
category_sales = pd.concat(
    category_results
)
```

```python
category_sales = (
    category_sales
    .groupby(level=0)
    .sum()
    .reset_index()
)
```

```python
category_sales.columns = [
    "Category",
    "Revenue"
]
```

```python
category_sales = category_sales.sort_values(
    "Revenue",
    ascending=False
)
```

```python
high_value_count = 0
```

```python
for chunk in pd.read_csv(
    FILE_PATH,
    chunksize=CHUNK_SIZE
):

    high_value_count += (
        chunk["Revenue"] > 100000
    ).sum()
```

```python
print(
    "High-value transactions:",
    high_value_count
)
```

```python
highest_transaction = None
```

```python
for chunk in pd.read_csv(
    FILE_PATH,
    chunksize=CHUNK_SIZE
):

    current_max = chunk["Revenue"].max()

    if (
        highest_transaction is None
        or current_max > highest_transaction
    ):
        highest_transaction = current_max
```

```python
print(
    "Highest transaction:",
    highest_transaction
)
```

```python
average_revenue = (
    total_revenue /
    total_transactions
)
```

```python
print(
    "Average transaction revenue:",
    round(average_revenue, 2)
)
```

```python
summary = pd.DataFrame({
    "Metric": [
        "Total Transactions",
        "Total Revenue",
        "Average Transaction Revenue",
        "Highest Transaction",
        "High Value Transactions"
    ],

    "Value": [
        total_transactions,
        total_revenue,
        round(average_revenue, 2),
        highest_transaction,
        high_value_count
    ]
})
```
